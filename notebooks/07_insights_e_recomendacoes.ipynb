## Nota metodológica geral

A partir da auditoria de robustez realizada em H4, identificou-se que os anos de 2008 a
2015 possuíam volume insuficiente de títulos (1 a 82 por ano) para sustentar qualquer
conclusão estatística — desvios padrão maiores que as próprias médias em diversos meses/
categorias. Esse mesmo filtro (`year_added >= 2016`, volume >= 100 títulos/ano) foi
replicado de forma consistente em H1, H2, H3, H5 e H6. A leitura de todas as hipóteses
temporais abaixo refere-se, portanto, ao período **2016–2020**, salvo indicação contrária.

### H4 — Sazonalidade de Adições

**Achado:** Não há um único mês de pico estatisticamente robusto. Embora dezembro tenha a
maior média (12,64%), esse valor é distorcido por um outlier isolado em 2016 (22,4%, mais
que o dobro dos outros 4 anos) — checagem de robustez confirmou que a diferença entre
dezembro e o 2º colocado (outubro, 10,51%) é menor que o próprio desvio padrão de dezembro.

**Achado mais robusto:** outubro é o mês que sustenta um padrão sazonal consistente — acima
da média teórica uniforme (8,33%) em todos os 5 anos observados (8,9% a 11,6%), com o menor
desvio padrão da série (1,28). Este é o indicador de sazonalidade mais confiável do estudo.

**Ressalva:** com apenas 5 anos de dado robusto, qualquer padrão sazonal aqui reportado deve
ser tratado como indicativo, não definitivo — recomenda-se revalidar quando mais anos de
dado íntegro estiverem disponíveis.

### H7 — Defasagem entre Lançamento e Adição

**Achado:** O comportamento típico de aquisição permaneceu estável entre 2016 e 2020 — a
mediana de defasagem se manteve constante em 1 ano durante todo o período de maior volume.

**Ressalva sobre a média:** a média subiu de ~3 para ~5,3 anos no mesmo período, efeito de
um pequeno volume de aquisições pontuais de conteúdo histórico/arquivístico (ex:
documentários de guerra de 1942–43, confirmados como registros legítimos na auditoria de
outliers). Este é o exemplo mais direto do projeto de por que a mediana foi adotada como
estatística de referência.

## Recomendações para os Stakeholders

### Para Mariana (Conteúdo)

- O catálogo é fortemente concentrado em Filmes (quase 70% do total), com crescimento de
  adições desacelerando em 2020 (-6,8% a.a. após 3 anos de forte expansão) — vale investigar
  se isso reflete mudança de estratégia interna ou limitação do próprio dataset (ano 2021
  parcial, fora desta análise).
- A diversidade de gêneros já atingiu sua cobertura máxima (42 categorias) desde 2016; o
  espaço de crescimento está mais em **concentração/diluição** entre gêneros existentes do
  que em ampliação do portfólio de categorias.
- A medição de retenção real, solicitada originalmente, segue fora de escopo por limitação
  da base — recomenda-se projeto de fase 2 com dados internos de consumo.

### Para Carlos (Marketing)

- Sazonalidade: não há um mês de pico isolado e robusto — a única evidência estatisticamente
  consistente é outubro, mês que historicamente concentra participação acima da média em
  todos os anos analisados (2016–2020). Recomenda-se cautela ao planejar campanhas
  fortemente ancoradas em dezembro, já que seu destaque é puxado por um único ano (2016),
  não por um padrão recorrente.
- Expansão internacional: a presença internacional já é majoritária (~63-69%) desde o
  início do período analisado, sem sinal de expansão adicional relevante — o catálogo já
  opera em patamar internacionalizado.
- Cadência de aquisição: expectativa realista de ~1 ano de defasagem entre lançamento e
  disponibilização na plataforma, estável desde 2016 — útil para calendário de campanhas
  de conteúdo recém-adquirido.
- Duração de filmes cresceu de forma consistente (83min → 102min), o que pode informar
  expectativa de formato em campanhas de novos lançamentos.

## Limitações Gerais do Estudo

- Dataset estático (corte em 25/09/2021), não reflete operação atual da Netflix.
- Ausência de métricas de consumo/audiência em toda a análise — nenhuma conclusão aqui
  mede sucesso ou desempenho de conteúdo, apenas composição da oferta.
- Todas as séries temporais (H1, H2, H3, H4, H5, H6) restritas a 2016–2020 por auditoria
  de robustez estatística — 2008–2015 excluídos por volume insuficiente.
- H4 não identifica pico sazonal único e definitivo — outubro é o indicador mais confiável,
  mas com apenas 5 anos de amostra robusta.
- H7 prioriza mediana sobre média, por distorção causada por outliers legítimos de
  conteúdo histórico.
